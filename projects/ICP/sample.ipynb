```python
# SMOKE TEST

import json
import time
from image import process_image

with open("pipeline_configs.json") as file:
    # load the configs to dictate how our pipeline works
    configs = json.load(file)

config = configs[1]
print("Running:", config["name"])

start = time.perf_counter()

for index in range(len(ds)):
    img_original = ds[index]["image"]
    img_processed = process_image(img_original, config)

    # the OCR step will go here later, using img_processed

    if (index + 1) % 1000 == 0:
        print(f"Processed {index + 1} of {len(ds)} images")

elapsed = time.perf_counter() - start
print(f"Processed {len(ds)} images in {elapsed:.1f} seconds")
```
